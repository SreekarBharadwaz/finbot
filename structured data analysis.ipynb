```python
# ============================================
# FINANCIAL RATIO ANALYSIS
# ============================================

def safe_divide(a, b):
    return np.where(
        (b != 0) & (~pd.isna(b)),
        a / b,
        np.nan
    )


# 1. Net Profit Margin
if "net_income" in financial_wide.columns and "revenue" in financial_wide.columns:
    financial_wide["net_profit_margin"] = (
        safe_divide(
            financial_wide["net_income"],
            financial_wide["revenue"]
        ) * 100
    )


# 2. Operating Margin
if "operating_income" in financial_wide.columns and "revenue" in financial_wide.columns:
    financial_wide["operating_margin"] = (
        safe_divide(
            financial_wide["operating_income"],
            financial_wide["revenue"]
        ) * 100
    )


# 3. Gross Margin
if "gross_profit" in financial_wide.columns and "revenue" in financial_wide.columns:
    financial_wide["gross_margin"] = (
        safe_divide(
            financial_wide["gross_profit"],
            financial_wide["revenue"]
        ) * 100
    )


# 4. Liability-to-Asset Ratio
if "liabilities" in financial_wide.columns and "assets" in financial_wide.columns:
    financial_wide["liability_asset_ratio"] = (
        safe_divide(
            financial_wide["liabilities"],
            financial_wide["assets"]
        )
    )


# 5. Cash-to-Asset Ratio
if "cash" in financial_wide.columns and "assets" in financial_wide.columns:
    financial_wide["cash_asset_ratio"] = (
        safe_divide(
            financial_wide["cash"],
            financial_wide["assets"]
        )
    )


display(financial_wide.head(10))
```

```python
# ============================================
# FINANCIAL TREND ANALYSIS
# ============================================

# Sort by company and year
financial_wide = financial_wide.sort_values(
    ["name", "year"]
).reset_index(drop=True)


# Metrics for trend analysis
trend_metrics = [
    "revenue",
    "net_income",
    "operating_income",
    "gross_profit",
    "assets",
    "liabilities",
    "cash",
    "eps_diluted"
]


# Calculate year-over-year growth
for metric in trend_metrics:

    if metric in financial_wide.columns:

        financial_wide[f"{metric}_growth_pct"] = (
            financial_wide
            .groupby("name")[metric]
            .pct_change() * 100
        )


# Display trend columns
trend_columns = [
    "name",
    "year"
]

for metric in trend_metrics:
    if f"{metric}_growth_pct" in financial_wide.columns:
        trend_columns.append(f"{metric}_growth_pct")


display(financial_wide[trend_columns].head(20))
```

```python
# ============================================
# FINANCIAL HEALTH SCORE
# ============================================

def calculate_health_score(row):

    score = 100

    # Profitability
    if "net_profit_margin" in row and pd.notna(row["net_profit_margin"]):
        if row["net_profit_margin"] < 0:
            score -= 25
        elif row["net_profit_margin"] < 5:
            score -= 10

    # Revenue growth
    if "revenue_growth_pct" in row and pd.notna(row["revenue_growth_pct"]):
        if row["revenue_growth_pct"] < 0:
            score -= 20

    # Net income growth
    if "net_income_growth_pct" in row and pd.notna(row["net_income_growth_pct"]):
        if row["net_income_growth_pct"] < 0:
            score -= 20

    # EPS growth
    if "eps_diluted_growth_pct" in row and pd.notna(row["eps_diluted_growth_pct"]):
        if row["eps_diluted_growth_pct"] < 0:
            score -= 15

    # Liability / asset ratio
    if "liability_asset_ratio" in row and pd.notna(row["liability_asset_ratio"]):
        if row["liability_asset_ratio"] > 0.75:
            score -= 15
        elif row["liability_asset_ratio"] > 0.60:
            score -= 5

    # Keep score between 0 and 100
    return max(0, min(100, score))


financial_wide["financial_health_score"] = (
    financial_wide.apply(calculate_health_score, axis=1)
)


# Health category
def health_category(score):

    if score >= 80:
        return "Strong"

    elif score >= 60:
        return "Moderate"

    elif score >= 40:
        return "Weak"

    else:
        return "Critical"


financial_wide["financial_health"] = (
    financial_wide["financial_health_score"]
    .apply(health_category)
)


display(
    financial_wide[
        [
            "name",
            "year",
            "financial_health_score",
            "financial_health"
        ]
    ].head(20)
)
```

```python
# ============================================
# FINANCIAL RISK DETECTION
# ============================================

def detect_risks(row):

    risks = []

    if row.get("revenue_growth_pct", 0) < 0:
        risks.append("Revenue Decline")

    if row.get("net_income_growth_pct", 0) < 0:
        risks.append("Net Income Decline")

    if row.get("eps_diluted_growth_pct", 0) < 0:
        risks.append("EPS Decline")

    if row.get("net_profit_margin", 100) < 5:
        risks.append("Low Profitability")

    if row.get("liability_asset_ratio", 0) > 0.75:
        risks.append("High Financial Leverage")

    if row.get("cash_asset_ratio", 1) < 0.05:
        risks.append("Low Cash Position")

    if len(risks) == 0:
        return "No Major Risk"

    return ", ".join(risks)


financial_wide["risk_flags"] = (
    financial_wide.apply(detect_risks, axis=1)
)


display(
    financial_wide[
        [
            "name",
            "year",
            "financial_health_score",
            "financial_health",
            "risk_flags"
        ]
    ].head(20)
)
```

```python
# ============================================
# FINANCIAL GUIDANCE GENERATION
# ============================================

def generate_guidance(row):

    guidance = []

    # Revenue
    revenue_growth = row.get("revenue_growth_pct")

    if pd.notna(revenue_growth):
        if revenue_growth > 10:
            guidance.append("Strong revenue growth")
        elif revenue_growth > 0:
            guidance.append("Revenue is growing moderately")
        else:
            guidance.append("Revenue decline requires attention")

    # Net income
    income_growth = row.get("net_income_growth_pct")

    if pd.notna(income_growth):
        if income_growth > 10:
            guidance.append("Profitability is improving")
        elif income_growth < 0:
            guidance.append("Profitability is declining")

    # EPS
    eps_growth = row.get("eps_diluted_growth_pct")

    if pd.notna(eps_growth):
        if eps_growth > 0:
            guidance.append("EPS shows positive momentum")
        else:
            guidance.append("EPS decline should be monitored")

    # Margin
    margin = row.get("net_profit_margin")

    if pd.notna(margin):
        if margin < 5:
            guidance.append("Improve cost efficiency and margins")

    # Leverage
    leverage = row.get("liability_asset_ratio")

    if pd.notna(leverage):
        if leverage > 0.75:
            guidance.append("Monitor debt and financial leverage")

    # Cash
    cash_ratio = row.get("cash_asset_ratio")

    if pd.notna(cash_ratio):
        if cash_ratio < 0.05:
            guidance.append("Strengthen liquidity position")

    if not guidance:
        return "Financial position appears stable"

    return "; ".join(guidance)


financial_wide["financial_guidance"] = (
    financial_wide.apply(generate_guidance, axis=1)
)


display(
    financial_wide[
        [
            "name",
            "year",
            "financial_health_score",
            "financial_health",
            "risk_flags",
            "financial_guidance"
        ]
    ].head(20)
)
```

```python
# ============================================================
# CREATE FINAL CHATBOT KNOWLEDGE FILE
# ============================================================

import json
import pandas as pd
import os

chatbot_records = []

# Use the final processed dataset
if "financial_final" in globals():

    df = financial_final.copy()

elif "financial_wide" in globals():

    df = financial_wide.copy()

else:

    raise ValueError(
        "financial_final or financial_wide is not available."
    )


# Convert every row into a chatbot-readable record
for _, row in df.iterrows():

    record = {}

    for column in df.columns:

        value = row[column]

        # Convert pandas/NumPy values to JSON-safe values
        if pd.isna(value):
            value = None

        elif hasattr(value, "item"):
            value = value.item()

        elif isinstance(value, (pd.Timestamp,)):
            value = str(value)

        else:
            value = str(value) if not isinstance(
                value,
                (int, float, bool)
            ) else value

        record[column] = value

    chatbot_records.append(record)


# Save JSON
json_path = os.path.join(
    CHATBOT_DIR,
    "chatbot_knowledge.json"
)

with open(
    json_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        chatbot_records,
        f,
        indent=2,
        ensure_ascii=False
    )


print("============================================")
print("CHATBOT KNOWLEDGE CREATED")
print("============================================")
print("Records:", len(chatbot_records))
print("File:", json_path)
```
